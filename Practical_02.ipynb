## List, Tuple, Set, Dictionary and String

**Subject:** Python for Data Science (BE05000231)

**Aim:** List, Tuple, Set, Dictionary and String.

```python
# List
numbers = [10, 20, 30, 40]
numbers.append(50)
numbers.remove(20)
print("List:", numbers)

# Tuple
student = ("Dharmik", 20, "Computer Engineering")
print("Tuple:", student)

# Set
values = {1, 2, 2, 3, 4}
values.add(5)
print("Set:", values)

# Dictionary
person = {"name":"Dharmik", "age":20, "branch":"Computer Engineering"}
person["city"] = "Surat"
print("Dictionary:", person)

# String
text = "Python for Data Science"
print(text.upper())
print(text.lower())
print(text.replace("Python", "PDS"))
print(text.split())
```
