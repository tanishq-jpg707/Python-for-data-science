## Conditional Statements, Loops and Loop Control

**Subject:** Python for Data Science (BE05000231)

**Aim:** Conditional Statements, Loops and Loop Control.

```python
# if, if-else, elif
marks = 72
if marks >= 40:
    print("Pass")
else:
    print("Fail")

if marks >= 75:
    print("Distinction")
elif marks >= 60:
    print("First Class")
else:
    print("Other")

# while loop
n = 1
while n <= 5:
    print(n, end=" ")
    n += 1
print()

# for loop and nested loop
for i in range(1, 4):
    for j in range(1, 4):
        print(i*j, end=" ")
    print()

# break, continue, pass
for i in range(1, 6):
    if i == 3:
        continue
    if i == 5:
        break
    print(i)

if True:
    pass
```
