## File Handling: Text, CSV and Binary Files

**Subject:** Python for Data Science (BE05000231)

**Aim:** File Handling: Text, CSV and Binary Files.

```python
# Text file: write, read and append
with open("sample.txt", "w") as f:
    f.write("Python for Data Science\n")
with open("sample.txt", "a") as f:
    f.write("Practical File Handling\n")
with open("sample.txt", "r") as f:
    print(f.read())

# CSV file
import csv
rows = [["Name", "Marks"], ["Dharmik", 85], ["Student2", 78]]
with open("students.csv", "w", newline="") as f:
    csv.writer(f).writerows(rows)
with open("students.csv", "r") as f:
    print(list(csv.reader(f)))

# Binary file
with open("sample.bin", "wb") as f:
    f.write(b"Python Binary Data")
with open("sample.bin", "rb") as f:
    print(f.read())
```
