## Pandas Series and DataFrame Operations

**Subject:** Python for Data Science (BE05000231)

**Aim:** Pandas Series and DataFrame Operations.

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40], name="Marks")
print("Series:
", s)

df = pd.DataFrame({
    "Name": ["Amit", "Riya", "Dharmik", "Neha"],
    "Department": ["CE", "CE", "CE", "IT"],
    "Marks": [75, 88, 92, 81]
})
print("DataFrame:
", df)
print("Selection:
", df[["Name", "Marks"]])
print("Filtering:
", df[df["Marks"] >= 80])
print("Grouping:
", df.groupby("Department")["Marks"].mean())
print("Aggregation:
", df.groupby("Department")["Marks"].agg(["mean", "max", "min"]))
```
