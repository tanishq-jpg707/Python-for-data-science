**Aim:** SciPy Stats: Measures and Statistical Tests.

```python
import numpy as np
from scipy import stats

marks = np.array([65, 70, 72, 75, 80, 85, 90, 92, 95])
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Mode:", stats.mode(marks, keepdims=True).mode[0])
print("Standard deviation:", np.std(marks, ddof=1))

# One-sample t-test
result = stats.ttest_1samp(marks, popmean=75)
print("T-statistic:", result.statistic)
print("P-value:", result.pvalue)
if result.pvalue < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```
