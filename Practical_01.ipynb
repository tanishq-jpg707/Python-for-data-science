**Aim:** Variables, Data Types and Operators.

```python
# Variables and data types
name = "Dharmik"
age = 20
percentage = 85.5
is_student = True

print("Name:", name)
print("Age:", age)
print("Percentage:", percentage)
print("Student:", is_student)
print(type(name), type(age), type(percentage), type(is_student))

# Operators
x, y = 10, 3
print("Arithmetic:", x+y, x-y, x*y, x/y, x%y, x//y, x**y)
x += 2
print("Assignment:", x)
print("Comparison:", x > y, x == y)
print("Logical:", x > 5 and y < 5, x > 20 or y < 5, not False)
print("Membership:", 3 in [1,2,3])
print("Identity:", x is y)
print("Bitwise:", 5 & 3, 5 | 3, 5 ^ 3, 5 << 1, 5 >> 1)
```
