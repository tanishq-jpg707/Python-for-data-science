### Dataset
This practical uses `employee.xlsx`, included in this ZIP package. Keep the Excel file in the same folder as this notebook.

```python
import pandas as pd

# Excel file is supplied with this practical package.
df = pd.read_excel("employee.xlsx")
print(df[["Age", "MonthlyIncome"]].head())

x = df["MonthlyIncome"]
print("Mean:", x.mean())
print("Median:", x.median())
print("Mode:", x.mode()[0])
print("Variance:", x.var())
print("Standard Deviation:", x.std())
print("Range:", x.max() - x.min())
print("Quartiles:")
print(x.quantile([0.25, 0.50, 0.75]))
print("IQR:", x.quantile(0.75) - x.quantile(0.25))

print("\nDescriptive summary:")
print(x.describe())
```
