### Dataset
This practical uses `employee.xlsx`, included in this ZIP package. Keep the Excel file in the same folder as this notebook.

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Data collection / loading
df = pd.read_excel("employee.xlsx")

# 2. Data understanding
print(df.head())
print(df.shape)
print(df.info())
print(df.describe())

# 3. Data preparation
print("Missing values:")
print(df.isnull().sum().sum())

# 4. Univariate analysis
plt.figure(figsize=(7,4))
plt.hist(df["Age"], bins=10)
plt.xlabel("Age")
plt.ylabel("Employees")
plt.title("Univariate Analysis - Age")
plt.show()

# 5. Bivariate analysis
plt.figure(figsize=(7,4))
plt.scatter(df["Age"], df["MonthlyIncome"], alpha=0.5)
plt.xlabel("Age")
plt.ylabel("Monthly Income")
plt.title("Bivariate Analysis - Age vs Monthly Income")
plt.show()

# 6. Multivariate analysis
selected = df[["Age", "MonthlyIncome", "JobSatisfaction", "YearsAtCompany"]]
print(selected.corr(numeric_only=True))

# 7. Conclusion
print("Data science lifecycle completed: collect -> understand -> prepare -> analyze -> visualize -> conclude")
```
