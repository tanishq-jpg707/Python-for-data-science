## User-defined Functions and Built-in Modules

**Subject:** Python for Data Science (BE05000231)

**Aim:** User-defined Functions and Built-in Modules.

```python
import math
import random
import statistics

def add(a, b):
    return a + b

def student_info(name, age=20):
    print("Name:", name, "Age:", age)

def total(*numbers):
    return sum(numbers)

def show_details(**data):
    print(data)

print("Addition:", add(10, 20))
student_info("Dharmik")
print("Total:", total(10, 20, 30))
show_details(name="Dharmik", branch="Computer Engineering")

values = [10, 20, 30, 40, 50]
print("Square root:", math.sqrt(25))
print("Random number:", random.randint(1, 10))
print("Mean:", statistics.mean(values))
print("Median:", statistics.median(values))
```
