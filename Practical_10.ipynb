## NumPy Array Creation, Indexing, Slicing, Reshaping, Broadcasting

**Subject:** Python for Data Science (BE05000231)

**Aim:** NumPy Array Creation, Indexing, Slicing, Reshaping, Broadcasting.

```python
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])
print("Array:
", arr)
print("Shape:", arr.shape)
print("Indexing:", arr[0, 1])
print("Slicing:
", arr[:, 1:])

reshaped = np.arange(1, 13).reshape(3, 4)
print("Reshaped:
", reshaped)

# Broadcasting
print("Broadcasting:
", arr + 10)
print("Multiplication:
", arr * 2)
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
```
